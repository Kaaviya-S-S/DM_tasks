```python
import pandas as pd
import numpy as np

file_name = '/content/drive/MyDrive/Data Mining Lab/Physical_infrastructure_facilities_in_Municipality_and_corporation_schools_in_Tamil_Nadu.xls'

df = pd.read_excel(file_name)
print(df)
```

```python
# How many attributes are present in the dataset?
print('No of attributes present in the dataset: ', df.shape[1])
```

```python
# Identify the type of the each attribute (Qualitative/Quantitative)
print('Data types of the attributes: \n', df.dtypes)

quantitative_cols = df.select_dtypes(include=[np.number,'datetime64']).columns.tolist()
print(f"\nQuantitative Columns: {quantitative_cols}")

qualitative_cols = df.select_dtypes(exclude=[np.number,'datetime64']).columns.tolist()
print(f"\nQualitative Columns: {qualitative_cols}")
```

```python
# If an attribute is identified as quantitative, what type of values can be associated with it?

quantitative_dtypes = set([df.dtypes[attribute] for attribute in quantitative_cols])
print('Quantitative datatypes: ', quantitative_dtypes)
```

```python
# If an attribute is identified as qualitative, what type of values can be associated with it?

qualitative_dtypes = set([df.dtypes[attribute] for attribute in qualitative_cols])
print('Qualitative datatypes: ', qualitative_dtypes)
```

```python
# How many districts, village panchayats and schools are included in the dataset?

no_districts = df['District'].nunique()
no_village_panchayat = df['Village Panchayat'].nunique()
no_schools = df['School_id'].nunique() # NOTE: considered school_id here, because school name is not unique!

print('No of Districts: ',no_districts)
print('No of Village Panchayat: ',no_village_panchayat)
print('No of Schools: ',no_schools)
```

```python
# List the village panchayats and the number of schools in each of them under each and every district.

schools_per_village_panchayat = df.groupby(['District', 'Village Panchayat'])['School_id'].nunique().reset_index()
schools_per_village_panchayat = schools_per_village_panchayat.rename(columns={'School_id': 'Number of Schools'})
print(schools_per_village_panchayat)
```

```python
# How many categories of schools are included and list their categories.

school_categories = df['Category of School'].unique()
print('Categories of schools: \n',school_categories)
print('No of categories of schools: ',len(school_categories))
```

```python
# Based on the availability of data which is/are the oldest school(s)?

# Check datatype
print(df['Yearof establishment'].dtype)

# drop the non-int values and NULL values
# Keep only pure 4-digit years
mask = df['Yearof establishment'].astype(str).str.fullmatch(r'\d{4}')

clean_df = df.loc[mask].copy()

# Convert year to int
clean_df['Yearof establishment'] = clean_df['Yearof establishment'].astype(int)

# Find oldest year
oldest_year = clean_df['Yearof establishment'].min()

print("Oldest year:", oldest_year)

# Print oldest school(s)
oldest_schools = clean_df[clean_df['Yearof establishment'] == oldest_year]

print("\nOldest school(s):")
print(oldest_schools[['School Name', 'Yearof establishment']])
```

```python
# Add another five interpretations on the dataset.

# which village panchayat(s) have the highest and lowest number of schools with count in the dataset?
schools_per_village_panchayat = df.groupby(['Village Panchayat'])['School_id'].nunique()
village_panchayat_with_highest_schools = schools_per_village_panchayat.idxmax()
village_panchayat_with_lowest_schools = schools_per_village_panchayat.idxmin()

print(f"1)\nVillage Panchayat with the highest number of schools: {village_panchayat_with_highest_schools} ({schools_per_village_panchayat.max()} schools)")
print(f"Village Panchayat with the lowest number of schools: {village_panchayat_with_lowest_schools} ({schools_per_village_panchayat.min()} schools)")

# ----------------------------------------------------------------------------------------------------------

# No of schools the data collection is completed?
no_of_schools_completed = (df['Status of the data set collected'] == 'Completed').sum()
print(f"\n2) No of schools the data collection is completed: {no_of_schools_completed}")

# ----------------------------------------------------------------------------------------------------------

# Which is the School with largest school area?
largest_school = df.loc[df['Total School area'].astype(float).idxmax()]
print(f"\n3) School with the largest school area: {largest_school['School Name']} ({str(largest_school['Total School area'])} acres)")

# ----------------------------------------------------------------------------------------------------------

# No of schools that offer more than one subject
no_of_schools_more_than_one_subject = (
    df['Subject offered']
    .dropna()                         # remove NULL / NaN
    .astype(str)
    .apply(lambda x: len(x.split(',')) > 1)
    .sum()
)
print("\n4) Number of schools offering more than one subject:", no_of_schools_more_than_one_subject)

# ----------------------------------------------------------------------------------------------------------

# Based on the availability of data which is/are the newest school(s)?
# Check datatype
print(df['Yearof establishment'].dtype)

# drop the non-int values and NULL values
# Keep only pure 4-digit years
mask = df['Yearof establishment'].astype(str).str.fullmatch(r'\d{4}')

clean_df = df.loc[mask].copy()

# Convert year to int
clean_df['Yearof establishment'] = clean_df['Yearof establishment'].astype(int)

# Find newest year
newest_year = clean_df['Yearof establishment'].max()

# Print newest school(s)
newest_schools = clean_df[clean_df['Yearof establishment'] == newest_year]

print("\n5) Newest school(s):")
print(newest_schools[['School Name', 'Yearof establishment']])
```
