```python
import pandas as pd

dataset = {
  "x": [10, 20, 30, 30, 5],
  "y": [2, 50, 10, 15, 10]
}
df = pd.DataFrame(dataset)

print(df)
```

```python
data = []
for i in range(len(df)):
  data.append([int(df.loc[i, "x"]), int(df.loc[i, "y"])])

print(data)
```

```python
import math
import numpy as np

def euclidean_distance(a, b):
  return math.sqrt((a[0] - b[0])**2 + (a[1] - b[1])**2)

distance_matrix = []
for i in range(len(data)):
  temp = []
  for j in range(len(data)):
    dist = euclidean_distance(data[i],data[j])
    temp.append(dist)
  distance_matrix.append(temp)

distance_matrix = np.matrix(distance_matrix)
print(distance_matrix)
```

**Interpretation**:

Most Similar: Obj3 and Obj4 are the most similar objects because they have the smallest Euclidean distance (5.00).

Least Similar: Obj1 and Obj2 are the most dissimilar because they have the largest distance (49.03)

```python
import numpy as np

print("Similarity matrix: S = 1/(1 + D)")
similarity_matrix = 1/(1 + distance_matrix)
print(similarity_matrix)
```

```python
from math import *
import numpy as np

def cosine_similarity(x,y):
  magnitude_x = sqrt(x[0]**2 + x[1]**2)
  magnitude_y = sqrt(y[0]**2 + y[1]**2)
  if magnitude_x == 0 or magnitude_y == 0:
    return 0.0
  similarity = (x[0]*y[0] + x[1]*y[1]) / (magnitude_x * magnitude_y)
  return similarity

print("Cosine Similarity: ")
cosine_similarity_matrix = []
for i in range(len(data)):
  temp = []
  for j in range(len(data)):
    similarity = cosine_similarity(data[i],data[j])
    temp.append(similarity)
  cosine_similarity_matrix.append(temp)


cosine_similarity_matrix = np.matrix(cosine_similarity_matrix)
print(cosine_similarity_matrix)
```

```python
import numpy as np

x = [data[i][0] for i in range(len(data))]
y = [data[i][1] for i in range(len(data))]

correlation = np.corrcoef(x, y)[0, 1]

print(f"Correlation coefficient between features x and y: {correlation:.4f}")
```
