```python
import pandas as pd

dataset = pd.read_csv('/content/drive/MyDrive/Data Mining Lab/RT_IOT2022.csv')
dataset.info()
```

```python
dataset.head()
```

```python
df = dataset.drop(columns=['Unnamed: 0'], axis=1)
target = dataset['Attack_type']

df.head()
```

1. Preprocess the data if duplication of instances are found/ drop the single valued columns

```python
# Dropping duplictes
df.drop_duplicates(inplace=True)

# dropping single values columns
single_value_columns = [col for col in df.columns if df[col].nunique() == 1]
print("Single values columns: ", single_value_columns)
df.drop(columns=single_value_columns, inplace=True)

df.head()
```

2. Identify the suitable normalization/discretization technique on features to transform the data

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd

# Identify categorical columns
categorical_cols = df.select_dtypes(include='object').columns
df_numeric = df.drop(columns=categorical_cols)

# Scale numeric features
scaler = StandardScaler()
df_scaled = pd.DataFrame(
    scaler.fit_transform(df_numeric),
    columns=df_numeric.columns
).reset_index(drop=True)
df_cat = df[categorical_cols].reset_index(drop=True)

df_scaled_concat = pd.concat([df_scaled, df_cat], axis=1)

# drop the target class
df_new = df_scaled_concat.drop(columns=['Attack_type'])
target = df_scaled_concat['Attack_type']
df_new.head()
```

3. Use Mutual Information and Correlation to select appropriate features. Do they yield similar results?

```python
# one hot encode the categorical values before finding MI and correlation

categorical_cols = df_new.select_dtypes(include='object').columns
df_encoded_cat = pd.get_dummies(df_new[categorical_cols], drop_first=False).reset_index(drop=True)

df_numeric = df_new.drop(columns=categorical_cols).reset_index(drop=True)
df_encoded = pd.concat([df_numeric, df_encoded_cat], axis=1)

df_encoded.head()
```

```python
'''
Mutual information measures the dependence between two variables.
High mutual information means that the features are highly dependent on the target variable.
'''
from sklearn.feature_selection import mutual_info_classif

# Calculate mutual information
mutual_info = mutual_info_classif(df_encoded, target)

# Display mutual information scores for each feature
mutual_info_df = pd.DataFrame({'Feature': df_encoded.columns, 'Mutual Information': mutual_info})
mutual_info_df.sort_values(by='Mutual Information', ascending=False, inplace=True)
print(mutual_info_df)
```

```python
'''
Correlation can help identify features that are highly correlated.
If two features are highly correlated, one can be dropped.
'''
# Calculate correlation matrix
correlation_matrix = df_encoded.corr().abs()

# Remove highly correlated features (threshold=0.95)
import numpy as np

threshold = 0.95
indices = np.where(correlation_matrix > threshold)
columns_to_drop = set([df_encoded.columns[i] for i, j in zip(*indices) if i < j])

print("Columns to drop: \n", columns_to_drop)

df_corr = df_encoded.drop(columns=columns_to_drop)
df_corr.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix

X = df_encoded
y = target

splits = [(0.7, 0.3), (0.8, 0.2), (0.9, 0.1)]

classifiers = {
    'Logistic Regression': LogisticRegression(),
    'Decision Tree': DecisionTreeClassifier(),
    'Naive Bayes': GaussianNB(),
    'SVM': SVC(),
    'K-Nearest Neighbors': KNeighborsClassifier()
}

def evaluate_model(classifier, X_train, X_test, y_train, y_test):
    model = classifier.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    print(classification_report(y_test, y_pred))
    print(confusion_matrix(y_test, y_pred))


for train_size, test_size in splits:
    print("===================================================================================================")
    print(f"Train-Test Split: {train_size*100}% - {test_size*100}%")
    X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=train_size, random_state=42)

    for name, clf in classifiers.items():
        print(f"\nEvaluating {name}...")
        evaluate_model(clf, X_train, X_test, y_train, y_test)
```

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris

iris = load_iris()
X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = pd.Series(iris.target)

df = X.copy()
df["target"] = y

df.head()
```

```python
# Remove duplicates
df = df.drop_duplicates()
df.head()
```

```python
# Remove single valued columns

single_valued_cols = [col for col in df.columns if df[col].nunique==1]
print("Single valued columns: ", single_valued_cols)
df = df.drop(columns=single_valued_cols)
df.head()
```

```python
# Separate Features and Target
X = df.drop("target", axis=1)
y = df["target"]
```

```python
# Normalization
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)
X_scaled.head()
```

```python
# Mutual Information
from sklearn.feature_selection import mutual_info_classif

mi_scores = mutual_info_classif(X_scaled, y)
mi_series = pd.Series(mi_scores, index=X.columns)

mi_series.sort_values(ascending=False)
```

```python
# coorelation
df_scaled = pd.concat([X_scaled, y], axis=1)

corr_matrix = df_scaled.corr().abs()
# print(corr_matrix)

threshold = 0.9
indices = np.where(corr_matrix < threshold)
columns_to_drop = set([df_scaled.columns[i] for i, j in zip(*indices) if i < j])

print("Columns to drop: \n", columns_to_drop)

# for highly correlated..
# indices = np.where(corr_matrix > threshold)
```

```python
# Train-test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3)  #70-30 split
```

```python
# Train models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

classifiers = {
    "Logistic Regression": LogisticRegression(),
    "Decision Tree": DecisionTreeClassifier(),
    "Naive Bayes": GaussianNB(),
    "SVM": SVC(),
    "K-Nearest Neighbors": KNeighborsClassifier()
}

for name, classifier in classifiers.items():
    classifier.fit(X_train, y_train)
```

```python
# 5-fold cross validation
from sklearn.model_selection import GridSearchCV

param_grid_lr = {
'C':[0.1,1,10],
'solver':['lbfgs','liblinear']
}

grid_lr = GridSearchCV(
LogisticRegression(max_iter=1000),
param_grid_lr,
cv=5
)

grid_lr.fit(X_train,y_train)

best_lr = grid_lr.best_estimator_
best_lr
```
