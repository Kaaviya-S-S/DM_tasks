# Age

```python
import matplotlib.pyplot as plt
import numpy as np

ages = [13, 40, 45, 16, 22, 25, 25, 22, 25, 25, 30, 52, 70, 33, 35, 15, 35, 36,33, 16, 19, 20, 35, 35, 46, 20, 21]

# Histogram
plt.hist(ages, color='skyblue', edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age distribution - Histogram')
plt.legend(['Age'])
plt.show()
```

```python
from matplotlib import pyplot as plt
import numpy as np
from collections import Counter

ages = [13, 40, 45, 16, 22, 25, 25, 22, 25, 25, 30, 52, 70, 33, 35, 15, 35, 36,33, 16, 19, 20, 35, 35, 46, 20, 21]

# Piechart
data = Counter(ages)
plt.figure(figsize=(6, 6))
plt.pie(list(data.values()), labels=list(data.keys()), autopct='%.0f%%')
plt.title('Age distribution - Piechart')
plt.show()
```

### Interpretations
- Dominant age group is 25 and 35 years
- Older age groups, such as 40-49, 50-59, and 70+, represent smaller segments of the total population, indicating fewer individuals in these categories
- Substantial portion of the data is concentrated in the younger to middle adult age ranges (20s and 30s).

# Iris

```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/Data Mining Lab/iris.csv', header=None)
headers = ["sepal length (cm)", "sepal width (cm)", "petal length (cm)", "petal width (cm)", "species"]
df.columns = headers
df.head()
```

```python
# a. Draw the scatter plots based on the feature sepal width and sepal length

# # **Using matplotlib
# import matplotlib.pyplot as plt
# import pandas as pd

# # Get unique species for coloring
# species = df['species'].unique()

# plt.figure(figsize=(8, 6))
# for s in species:
#     subset = df[df['species'] == s]
#     plt.scatter(subset['sepal length (cm)'], subset['sepal width (cm)'], label=s)

# plt.xlabel('sepal length (cm)')
# plt.ylabel('sepal width (cm)')
# plt.title('Scatter plot of sepal length and sepal width')
# plt.legend()
# plt.show()

# **Using seaborn
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df,
    x='sepal length (cm)',
    y='sepal width (cm)',
    hue='species'
)

plt.title('Scatter plot of sepal length and sepal width')
plt.xlabel('sepal length (cm)')
plt.ylabel('sepal width (cm)')
plt.legend(title='Species')
plt.show()
```

```python
# b. Identify the suitable visualization and generate it to compare the petal length of various species

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
sns.boxplot(x="species", y='petal length (cm)', data=df, hue='species')
plt.title('Boxplot of petal length for each species')
plt.show()
```

```python
# c. Create swarm plots for petal width and sepal width and interpret the following:
#   i. What is the spread or the range of the measurements?
#   ii. Identify the sepal and petal width having the highest frequency

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
sns.swarmplot(x="species", y='petal width (cm)', data=df, hue='species')
plt.title('Swarm plot of petal width for each species')
plt.grid(True)

plt.figure(figsize=(6,6))
sns.swarmplot(x="species", y='sepal width (cm)', data=df, hue='species')
plt.title('Swarm plot of sepal width for each species')
plt.grid(True)

plt.show()
```

#### i) What is the spread or the range of the measurements?

petal width:
*   setosa -> 0.1cm - 0.6cm
*   versicolor -> 1cm - 1.8cm
*   virginica -> 1.4cm - 2.5cm

sepal width:
*   setosa -> 2.3cm - 4.4cm
*   versicolor -> 2cm - 3.4cm
*   virginica -> 2.2cm - 3.8cm

---

#### ii) Identify the sepal and petal width having the highest frequency

In a swarm plot, the value where points are most densely packed vertically indicates the highest frequency.
These values appear as the thickest / most crowded horizontal bands for each species.

-> petal width:
*   setosa - ~0.2cm
*   versicolor - ~1.3cm
*   virginica - ~1.8cm

-> sepal width:
*   setosa - ~3.4cm
*   versicolor - ~3cm
*   virginica - ~3cm


```python
# d. Generate swarm plot based on sepal length according to various categories of the species

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
sns.swarmplot(x="species", y='sepal length (cm)', data=df, hue='species')
plt.title('Swarm plot of sepal length for each species')
plt.grid(True)
plt.show()
```

```python
# e. Compare the histogram and density plots for sepal width

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
sns.histplot(data=df, x='sepal width (cm)', hue='species')
plt.title('Histogram of sepal width for each species')

plt.figure(figsize=(6,6))
sns.kdeplot(data=df, x='sepal width (cm)', hue='species')
plt.title('Density plot(KDE) of sepal width for each species')
plt.show()

```

```python
# 3. Plot the box plots of all the features, excluding the class, in a single frame

import seaborn as sns
import matplotlib.pyplot as plt

# Exclude the class column
features = df.drop(columns=['species'])

plt.figure(figsize=(8,6))
sns.boxplot(features)
plt.title("Box plots of Iris features (excluding class)")
plt.ylabel("Value (cm)")
plt.show()
```

```python
# 4. Analyse the distribution shape of the petal length using violin plot

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
sns.violinplot(x="species", y='petal length (cm)', data=df, hue='species')
plt.title('Violin plot of petal length for each species')
plt.grid(True)
plt.show()
```

### Interpretation on violin plot

#### i) Distribution
- Setosa - Normal distribution (since it is bulged in the middle)
- versicolor -  Normal distribution but Slightly negatively skewed (since it is heavy at the top)
- virginica -  Normal distribution but Slightly positively skewed ( since it is heavy at the bottom)


#### ii) Comparing groups
 Setosa in general has the lower petal length while virginica has the higher petal length (since once group is positioned below the other vertically)

#### iii) Density
Most common petal length for each species
- setosa is ~1.5cm
- versicolor is ~4.6cm
- virginica is ~5.5cm
