# Automobile spares

```python
import pandas as pd

fileName = '/content/drive/MyDrive/Data Mining Lab/Automobile spare.csv'
df = pd.read_csv(fileName, header=None)
print(df)
```

```python
data = df.values[0]
display(data)
```

```python
# What is the maximum number of spares sold?
max_num_spares = max(data)
print(f"The maximum number of spares sold: {max_num_spares}")

# And on how many days the maximum sales have been achieved?
num_days_max_sales = sum(data==max_num_spares)
print(f"On {num_days_max_sales} days the maximum sales have been achieved.")
```

```python
# Find the number of days having no sales.
num_days_no_sales = sum(data==0)
print(f"No. of days having no sales: {num_days_no_sales}")
```

```python
# How many spares have been sold during the first half, i.e. first 30 days and second half of the period and which half produced the highest sales?

first_half_data = data[:30]
second_half_data = data[30:]

first_half_sales = sum(first_half_data)
second_half_sales = sum(second_half_data)

print(f"No. of spares sold during first half (30 days): {first_half_sales}")
print(f"No. of spares sold during second half (30 days): {second_half_sales}")

highest_sales_half = "First Half" if first_half_sales > second_half_sales else "Second Half"
print(f"\nHigest sales is during '{highest_sales_half}'")

# In each half, what is the maximum sales achieved?
print(f"\nMax sales in first half: {max(first_half_data)}")
print(f"Max sales in second half: {max(second_half_data)}")
```

# Age

```python
import numpy as np
from collections import Counter
import matplotlib.pyplot as plt

ages = [13, 40, 45, 16, 22, 25, 25, 22, 25, 25, 30, 52, 70, 33, 35, 15, 35, 36,33, 16, 19, 20, 35, 35, 46, 20, 21]

# What is the mean of the data? What is the median?
# What is the mode of the data?

mean_ages = np.mean(ages)
median_ages = np.median(ages)

freq_counter = Counter(ages)
max_count = max(freq_counter.values())
mode_ages = [age for age,freq in freq_counter.items() if freq==max_count]

print(f"Mean of ages: {mean_ages}")
print(f"Median of ages: {median_ages}")
print(f"Mode of ages: {mode_ages}")
```

```python
# What is the midrange of the data?

# midrange = (maxVal + minVal)/2
midrange_age = (max(ages)+min(ages))/2
print(f"Midrange of ages: {mean_ages}")
```

```python
# Find the first quartile (Q1) and the third quartile (Q3) of the data.

# quartile
# # Q1 Position: (n + 1) * (1/4)
# # Q2 Position: (n + 1) * (1/2)
# # Q3 Position: (n + 1) * (3/4)
# sorted_ages = sorted(ages)
# n = len(ages)
# Q1_position = int((n+1)*(1/4))
# Q3_position = int((n+1)*(3/4))
# print("\nusing formula")
# print(f"Q1: {sorted_ages[Q1_position]}")
# print(f"Q3: {sorted_ages[Q3_position]}")

Q1 = np.percentile(ages, 25)
Q3 = np.percentile(ages, 75)
print("using numpy - percentile")
print(f"Q1: {Q1}")
print(f"Q3: {Q3}")
```

```python
# Draw a boxplot of the data and interpret the box plot.

plt.boxplot(ages, vert=False)
plt.title('Box Plot - Ages')
plt.xlabel('Ages (Years)')
plt.ylabel('Age Distribution')
plt.show()
```

#Statewise secondary

```python
import pandas as pd

filename = '/content/drive/MyDrive/Data Mining Lab/2015_16_Statewise_Secondary.csv - 2015_16_Statewise_Secondary.csv.csv'
df = pd.read_csv(filename)
print(df)
```

```python
# Are there missing values in the dataset? If yes, on which attributes?

missing_cols = df.columns[df.isnull().any()]
print(missing_cols)
```

```python
# How many states are listed in the dataset?

no_states = len(df['statcd'])
print(f"No. of states: {no_states}")
```

```python
# Which state(s) possess the highest and lowest literacy rate?

highest_literacy_rate_value = df['literacy_rate'].max()
lowest_literacy_rate_value = df['literacy_rate'].min()

states_highest_literacy = df[df['literacy_rate'] == highest_literacy_rate_value]['statname'].tolist()
states_lowest_literacy = df[df['literacy_rate'] == lowest_literacy_rate_value]['statname'].tolist()

print(f"States with Highest literacy rate ({highest_literacy_rate_value}): {', '.join(states_highest_literacy)}")
print(f"States with Lowest literacy rate ({lowest_literacy_rate_value}): {', '.join(states_lowest_literacy)}")
```

```python
# How much percentage of schools have a single teacher in secondary level?
# sing_tch_sch_5 - Schools with Single Teacher: Secondary only
# sing_tch_sch_all - Schools with Single Teacher: All Schools

schools_single_teacher_secondary_level = df['sing_tch_sch_5'].sum()
total_schools_single_teacher = df['sing_tch_sch_all'].sum()
percentage_single_teacher = (schools_single_teacher_secondary_level/total_schools_single_teacher)*100
print(f"Percentage of schools with single teacher in secondary level: {percentage_single_teacher:.2f}%")
```

```python
# How many categories of schools are included in the dataset? List them.
import re

count = 0
pattern = r'^sch_\d+$'
school_categories = []

for col in df.columns:
    if re.search(pattern, col):
        school_categories.append(col)
        count += 1

print(f"School categories: {school_categories}")
print(f"No. of school categories: {count}")
```
